### <span style="color:red"> Key Differences: np.array() vs np.asarray() </span>

```python
# 1. COPYING BEHAVIOR - The main difference
# np.array() ALWAYS creates a new copy
# np.asarray() creates a copy ONLY if needed

# Example with a regular Python list
my_list = [1, 2, 3, 4, 5]

# Both create arrays from the list (both make copies since input is not already an ndarray)
arr1 = np.array(my_list)
arr2 = np.asarray(my_list)

print("Original list:", my_list)
print("np.array result:", arr1)
print("np.asarray result:", arr2)
print("Are they equal?", np.array_equal(arr1, arr2))
```

```python
alist = [1,2,3,4,5,6]

blist = [[1,2], [3,4]]

arr2 = np.array(alist)
arr1 =  np.asarray(blist, order='F')
```

```python
print(arr2)
print(arr1)
print(arr1.shape)
```

```python
# 2. THE KEY DIFFERENCE: When input is already a NumPy array
existing_array = np.array([10, 20, 30, 40])

# np.array() creates a NEW copy
new_copy = np.array(existing_array)

# np.asarray() returns the SAME array (no copy if not needed)
same_array = np.asarray(existing_array)

print("Original array:", existing_array)
print("New copy from np.array():", new_copy)
print("Same array from np.asarray():", same_array)

# Test if they are the same object in memory
print("\nMemory comparison:")
print("existing_array is new_copy:", existing_array is new_copy)
print("existing_array is same_array:", existing_array is same_array)

# Let's modify the original and see what happens
existing_array[0] = 999

print("\nAfter modifying original array:")
print("Original array:", existing_array)
print("New copy (unchanged):", new_copy)
print("Same array (changed too!):", same_array)
```

```python
# 3. Performance implications
import time

# Create a large array
large_array = np.random.rand(1000000)

# Time np.array() - always copies
start = time.time()
for _ in range(100):
    copy_result = np.array(large_array)
array_time = time.time() - start

# Time np.asarray() - no copy needed
start = time.time()
for _ in range(100):
    asarray_result = np.asarray(large_array)
asarray_time = time.time() - start

print(f"Time for 100 np.array() calls: {array_time:.4f} seconds")
print(f"Time for 100 np.asarray() calls: {asarray_time:.4f} seconds")
print(f"np.asarray() is {array_time/asarray_time:.1f}x faster when no copy is needed!")
```

```python
# 4. When np.asarray() DOES create a copy
# Case 1: Different dtype
int_array = np.array([1, 2, 3, 4], dtype=int)
float_array = np.asarray(int_array, dtype=float)  # Forces a copy due to dtype change

print("Original int array:", int_array)
print("Float array from asarray:", float_array)
print("Are they the same object?", int_array is float_array)

# Case 2: Different order (C vs Fortran)
c_array = np.array([[1, 2], [3, 4]], order='C')
f_array = np.asarray(c_array, order='F')  # Forces a copy due to order change

print("\nOriginal C-order array:", c_array)
print("Fortran-order array:", f_array)
print("Are they the same object?", c_array is f_array)
```

### <span style="color:green"> Summary: When to use which? </span>

**Use `np.array()`:**
- When you **always want a copy** (safe for modifications)
- When you want to **ensure data independence**
- When working with **mixed data types** or **converting data types**

**Use `np.asarray()`:**
- When you want **maximum efficiency** (avoid unnecessary copies)
- When you **don't mind sharing memory** with the original
- In **function parameters** where input might already be an array
- When you want to **preserve the original if it's already a compatible array**

**Key Rule:** `np.asarray()` is **more efficient** but **potentially less safe**, while `np.array()` is **always safe** but **potentially less efficient**.

## <span style="color:blue"> Pandas </span>

Pandas is an open-source Python library providing efficient, easy-to-use data structure and data analysis tools. The name Pandas is derived from "Panel Data" - an Econometrics from Multidimensional Data. Pandas is well suited for many different kinds of data:
1. Tabular data with heterogeneously-type columns.
2. Ordered and unordered time series data.
3. Arbitary matrix data with row and column labels.
4. Any other form observational/statistical data sets. The data actually need not be labeled at all to be placed into a pandas data structure.

Pandas provides three data structure - all of which are build on top of the NumPy array - all the data structures are value-mutable
1. Series (1D) - labeled, homogenous array of immutable size
2. DataFrames (2D) - labeled, heterogeneously typed, size-mutable tabular data structures
3. Panels (3D) - Labeled, size-mutable array

```python
import pandas as pd
```

### <span style="color:orange"> Series </span>

1. A Series is a single-dimensional array structures that stores homogenous data i.e., data of a single type.
2. All the elements of a Series are value-mutable and size-immutable
3. Data can be of multiple data types such as ndarray, lists, constants, series, dict etc.
4. Indexes must be unique, hashable and have the same length as data. Defaults to np.arrange(n) if no index is passed.
5. Data type of each column; if none is mentioned, it will be inferred; automatically
6. Deep copies data, set to false as default

### <span style="color:orange"> DataFrames </span>

1. A DataFrame is a 2D data structure in which data is aligned in a tabular fashion consisting of rows & columns
2. A DataFrame can be created using the following constructor - pandas.DataFrame(data, index, dtype, copy)
3. Data can be of multiple data types such as ndarray, list, constants, series, dict etc.
4. Index Row and column labels of the dataframe; defaults to np.arrange(n) if no index is passed
5. Data type of each column
6. Creates a deep copy of the data, set to false as default

### <span style="color:orange"> Creating a DataFrame </span>

```python
# Converting a list into a DataFrame
list1 = [10, 20, 30, 40]
table = pd.DataFrame(list1)
print(table)
```

### <span style="color:orange"> Plotting in Matplotlib </span>

```python
plt.plot([1,2,3,4]) # List of vertical co-ordinates of the points plotted
plt.show() # Displays plot
# Implicit X-axis values from 0 to (N-1) where N is the length of the list
```

```python
# We can specify the values for both axes
x = range(5) # Sequence of values for the x-axis
# X-axis values specified - [0,1,2,3,4]
plt.plot(x, [x1**2 for x1 in x]) # vertical co-ordinates of the points plotted: y = x^2
plt.show()
```

```python
# The scale of the plot can also be set using xlim() and ylim()
x = range(5)
plt.plot(x, [x1 for x1 in x],
         x, [x1*2 for x1 in x],
         x, [x1*4 for x1 in x])
plt.grid(True)
plt.xlim(-1, 5)
plt.ylim(-1, 10)
plt.show()
```

### <span style="color:orange"> Styling </span>

```python
# Matplotlib allows to choose custom colours for plots
y = np.arange(1, 3)
plt.plot(y, 'y') # Specifying line colours
plt.plot(y+5, 'm')
plt.plot(y+10, 'c')
plt.show()
```

Color code:
1. b = Blue
2. c = Cyan
3. g = Green
4. k = Black
5. m = Magenta
6. r = Red
7. w = White
8. y = Yellow

```python
# Matplotlib allows different line styles for plots
y = np.arange(1, 100)
plt.plot(y, '--', y*5, '-.', y*10, ':')
plt.show()
# - Solid line
# -- Dashed line
# -. Dash-Dot line
# : Dotted Line
```

```python
# Matplotlib provides customization options for markers
y = np.arange(1, 3, 0.2)
plt.plot(y, '*',
        y+0.5, 'o',
        y+1, 'D',
        y+2, '^',
        y+3, 's') # Specifying line styling
plt.show()
```
